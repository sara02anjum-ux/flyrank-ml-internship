## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

Every recommended action should receive human review before any content change is made.

The reviewer should check the page, its purpose, recent changes, business context, and whether the recommended action makes sense.

### Human review rules

* Review the evidence behind the recommendation.
* Confirm that the page is actually relevant for the suggested action.
* Check recent content or performance changes.
* Record the final decision and reason.

### No-go cases

The system should NOT automatically:

* Rewrite or delete content.
* Change titles or metadata without review.
* Declare a page as definitely declining.
* Make claims about search-engine causality.
* Publish content changes automatically.
* Override an experienced human reviewer's decision.
