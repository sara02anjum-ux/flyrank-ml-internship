# Predicting Search-Performance Decline for Content Prioritization

## Abstract

This study asks whether historical page-level search-performance signals can support the prioritization of pages that may require review. The project constructs a transparent baseline ranking and investigates a Decision Tree classification approach for identifying pages associated with declining performance. The baseline ranks 30,000 anonymized page records using search demand, impressions, CTR opportunity, content age, and average position, while the modeling workflow defines `is_declining_label` as the target and excludes target-defining fields from the feature set. The available project artifacts establish the baseline queue and leakage controls, but the Decision Tree training and final model-versus-baseline evaluation were not successfully reproduced in the available notebook run, so no unsupported model performance number is reported. The resulting artifact is therefore framed as directional decision support for prioritizing human review rather than as a guaranteed predictor of future search performance.

---

## 1. Introduction / Problem Statement

Content teams may have many pages that could potentially benefit from review, while the resources available for detailed investigation are limited. A practical question is therefore whether available search-performance signals can be used to create a ranked review queue.

This project investigates whether page-level historical signals can support that prioritization. The intended decision is not to automatically change content, but to help a reviewer decide which pages deserve attention first.

The analysis emphasizes transparent scoring, leakage awareness, validation discipline, and cautious interpretation of model outputs.

## 2. Data

### Dataset

The analysis uses the anonymized FlyRank ML Internship dataset available in the project repository. The modeling file contains **30,000 page-level records and 44 columns**.

The available fields include search-demand, traffic, engagement, content-age, freshness, ranking, and trend-related signals.

### Baseline Scope

The Week-4 baseline ranks all 30,000 records using five weighted signals:

| Signal             | Weight |
| ------------------ | -----: |
| Search volume      |    30% |
| 90-day impressions |    25% |
| CTR opportunity    |    20% |
| Content age        |    15% |
| Average position   |    10% |

The resulting ranked queue is written to `work/outputs/baseline_action_score.csv`.

### Exclusions and Public Safety

The modeling workflow excludes fields that directly define or reveal the target. In particular:

* `is_declining_label`
* `trend_direction`
* `trend_pct`

are treated as leakage-risk fields and should not be used as predictive features.

Private client information, private queries, credentials, and other non-public information are not included in the research artifact.

The available notebook evidence does not establish a specific dataset release identifier or exact calendar date window, so these details are not inferred or invented.

## 3. Methodology

### Research Setup

The task is treated as a supervised classification and ranking problem. The intended target is `is_declining_label`, representing whether a page meets the project's predefined decline criterion.

### Baseline

The Week-4 baseline is a transparent hand-rule score:

* 30% search-volume rank
* 25% 90-day-impression rank
* 20% inverse CTR rank
* 15% content-age rank
* 10% inverse average-position rank

All 30,000 pages are ranked by this score.

### Model

The planned machine-learning model is a Decision Tree classifier. A Decision Tree is appropriate for this exploratory task because the target is binary and the model can represent non-linear relationships while remaining relatively interpretable.

### Validation

The intended comparison is between the learned model and the Week-4 baseline under the same evaluation setup. The validation work also identifies grouped/client-level validation as a more conservative check because a random row split may allow related records from the same client to appear in both training and evaluation data.

### Leakage Checks

The leakage audit identifies `is_declining_label`, `trend_direction`, and `trend_pct` as fields that must be excluded from the final feature set.

The Week-4 baseline also checked for future-window and product-related fields. The reported audit found:

* Future-window columns: none
* Product-flag columns: none

These checks reduce the risk of using information that directly reveals the target or belongs to a future observation period.
