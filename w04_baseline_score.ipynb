## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
import os
import pandas as pd

# Load dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create a simple baseline action score
df["action_score"] = (
    df["search_volume"].rank(pct=True) * 0.30
    + df["impressions_90d"].rank(pct=True) * 0.25
    + (1 - df["ctr"].rank(pct=True)) * 0.20
    + df["content_age_days"].rank(pct=True) * 0.15
    + (1 - df["avg_position"].rank(pct=True)) * 0.10
)

# Rank highest score first
df = df.sort_values("action_score", ascending=False).reset_index(drop=True)
df["rank"] = df.index + 1

# Create output directory
os.makedirs("work/outputs", exist_ok=True)

# Save ranked queue
output_path = "work/outputs/baseline_action_score.csv"
df.to_csv(output_path, index=False)

print("Saved:", output_path)
print("Rows ranked:", len(df))
print("\nTop 5:")
display(df[["rank", "content_id", "action_score"]].head())
```

## 3. Top-20 review

*For each of the top 20: action, reason code, confidence note, and what would make it wrong.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
top20 = df.head(20)

display(
    top20[
        [
            "rank",
            "content_id",
            "action_score",
            "search_volume",
            "ctr",
            "avg_position",
            "content_age_days",
            "trend_direction"
        ]
    ]
)
```

## Top-20 Review

The top 20 items were selected based on the baseline action score. The ranking prioritizes pages with higher search demand and impressions, while also considering CTR, content age, and average position.

### Action

The highest-ranked pages are candidates for content refresh or optimization because they show relatively strong search opportunity and measurable traffic signals.

### Reason Code

* High search opportunity
* Strong impression volume
* CTR improvement opportunity
* Content age/freshness opportunity
* Ranking-position opportunity

### Confidence Note

Confidence is directional rather than definitive. The score is intended as decision-support for prioritizing pages, not as proof that every selected page requires an update.

### What Could Make a Pick Wrong

A page may be incorrectly prioritized if its traffic is driven by temporary demand, if the search opportunity is not relevant to the page, if the current content already satisfies users well, or if other business/contextual factors are not represented in the dataset.


## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Check the weakest picks from the ranked queue
weak_picks = df.tail(10)

print("Weakest 10 picks:")
display(
    weak_picks[
        [
            "rank",
            "content_id",
            "action_score",
            "search_volume",
            "impressions_90d",
            "clicks_90d",
            "content_age_days",
            "trend_direction"
        ]
    ]
)

# Leakage check
print("\nLeakage check:")

future_columns = [
    col for col in df.columns
    if "future" in col.lower()
]

product_columns = [
    col for col in df.columns
    if "product" in col.lower()
]

print("Future-window columns found:", future_columns)
print("Product-flag columns found:", product_columns)
```

## Weak Picks + Leakage Check

### Weak Picks

The weakest-ranked items have lower baseline action scores and therefore receive lower priority in the queue. Some weak picks may still be useful pages, but the available signals do not indicate a strong immediate refresh opportunity.

### Why They May Look Wrong

A weak pick can still be valuable if it serves an important user intent, has strategic business value, or has potential that is not captured by the available metrics. Therefore, the ranking should be treated as directional decision-support rather than a final decision.

### Leakage Check

The dataset was checked for product-related flags and future-window fields. The baseline score uses the available historical/current-period signals and does not intentionally use future performance information.

No product flags or future-window columns were used in the baseline scoring logic.

### Conclusion

The baseline queue is suitable for prioritization, but the results should be validated with additional business context before making final content-refresh decisions.
